```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn import svm
from sklearn.svm import SVC, LinearSVC
```

Sebagai percobaan gunakan data dengan pemanggilan data sebagai berikut

```python
data = pd.read_csv('bill_authentication.csv')
data.head()
```

pecah data menjadi data atribut dan data kelas

```python
X = data.drop('Class', axis=1)
Y = data['Class']
```

gunakan splitting untuk membagi data tersebut dalam data training dan data testing

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.20)
```

Gunakan metode fit pada SVC untuk membentuk model SVM dari data training

```python
modelSVM = svm.SVC(kernel='linear')
modelSVM.fit(X_train, Y_train)
```

gunakan model SVM dari hasill data training untuk membuat prediksi dari data

```python
Y_pred = modelSVM.predict(X_test)
Y_pred
```

Jalankan confusion matrix, precision, recall dan F1 untuk mengevaluasi hasil klasifikasi.

```python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
print(confusion_matrix(Y_test,Y_pred))
print(classification_report(Y_test,Y_pred))
print('akurasi : ', accuracy_score(Y_test, Y_pred))
```

Implementasi kernel pada SVM

```python
b = pd.read_csv('iris-data-prak.csv')
colnames = ['sepal-length', 'sepal-width', 'petal-length', 'petal-width', 'Class']
irisdata = pd.read_csv('iris-data-prak.csv', names=colnames, delimiter=';', skiprows=2)
irisdata.head()
```

Memecah data

```python
irisdata_cleaned = irisdata.dropna()
X = irisdata_cleaned.drop('Class', axis=1)
Y = irisdata_cleaned['Class']
```

membagi data

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.20)
```

gunakan kernel polinomial

```python
SVMPoli = svm.SVC(kernel='poly', degree=8)
SVMPoli.fit(X_train, Y_train)

Y_pred = SVMPoli.predict(X_test)
Y_pred

print(confusion_matrix(Y_test,Y_pred))
print(classification_report(Y_test,Y_pred))
print('akurasi : ', accuracy_score(Y_test, Y_pred))
```

Berikut untuk kernel gaussian

```python
SVMPoli = svm.SVC(kernel='rbf')
SVMPoli.fit(X_train, Y_train)

Y_pred = SVMPoli.predict(X_test)
Y_pred

print(confusion_matrix(Y_test,Y_pred))
print(classification_report(Y_test,Y_pred))
print('akurasi : ', accuracy_score(Y_test, Y_pred))
```

Sedangkan untuk kernel sigmoid

```python
SVMPoli = svm.SVC(kernel='sigmoid')
SVMPoli.fit(X_train, Y_train)

Y_pred = SVMPoli.predict(X_test)
Y_pred

print(confusion_matrix(Y_test,Y_pred))
print(classification_report(Y_test,Y_pred))
print('akurasi : ', accuracy_score(Y_test, Y_pred))
```
