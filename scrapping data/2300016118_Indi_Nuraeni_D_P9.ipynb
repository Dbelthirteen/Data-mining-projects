```python
import tweepy
import pandas as pd
import datetime

BEARER_TOKEN = "AAAAAAAAAAAAAAAAAAAAAAun6AEAAAAAlYpKGqZJJHflIixRIkcAfYwJnVY%3DwZHX8tIianjWeWa1VJMRFkRkg2cfi3Bw2DYVT1SGcIfSyW1cOl"

client = tweepy.Client(bearer_token=BEARER_TOKEN)

def search_tweets(keyword, max_results=100):
    """
    Mencari tweet berdasarkan keyword
    """
    try:
        query = f'{keyword} lang:id -is:retweet'

        tweets = client.search_recent_tweets(
            query=query,
            max_results=max_results,
            tweet_fields=['created_at', 'public_metrics', 'author_id', 'text'],
            user_fields=['username', 'name'],
            expansions=['author_id']
        )

        data = []
        users = {user.id: user for user in tweets.includes['users']}

        for tweet in tweets.data:
            user = users[tweet.author_id]
            data.append({
                'id': tweet.id,
                'created_at': tweet.created_at,
                'text': tweet.text,
                'author': user.username,
                'author_name': user.name,
                'retweets': tweet.public_metrics['retweet_count'],
                'likes': tweet.public_metrics['like_count'],
                'replies': tweet.public_metrics['reply_count']
            })

        return pd.DataFrame(data)

    except Exception as e:
        print(f"Error: {e}")
        return pd.DataFrame()

if __name__ == "__main__":
    df_tweets = search_tweets("pemilu 2024", max_results=50)

    if not df_tweets.empty:
        df_tweets.to_csv('tweets_pemilu.csv', index=False)
        print(f"Berhasil mengumpulkan {len(df_tweets)} tweet")
        print(df_tweets.head())
```
