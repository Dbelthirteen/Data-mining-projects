**Impor library Numpy, Pandas dan Apyori**

```python
import numpy as np
import pandas as pd
from apyori import apriori
```

**Tambahkan file Assignment-1_Data.xlsx**

```python
from google.colab import files
uploaded = files.upload()
```

**Panggil data yang telah ditambahkan**

```python
data = pd.read_excel('Assignment-1_Data.xlsx')
data.head()
```

**Cek Ukuran data**

```python
data.shape
```

**Mengambil isi data dan menjadikan dalam bentuk tipe data list**

```python
records = []
for i in range(0, 20):
  records.append([str(data.values[i,j]) for j in range(0, 6)])
```

```python
print(records)
```

**Jalankan metode apriori untuk mendapatkan rule**

```python
association_rules = apriori(records,
                            min_support=0.1,
                            min_confidence=0.2,
                            min_lift=1.0,
                            min_length=2)
association_results = list(association_rules)
```

**Lihat rule yang terbentuk**

```python
assoc_rule = list(association_results)
print(assoc_rule)
```

**Lihat banyaknya rule yang terbentuk**

```python
print(len(assoc_rule))
```

**Rule yang terbentuk lebih detail**

```python
for item in assoc_rule:
    pair = item[0]
    items = [x for x in pair]

    if len(items) > 1:
        print("Rule: " + items[0] + " -> " + items[1])
        print("Support: " + str(item[1]))
```

```python
print("confidence: " + str(item[2][0][2]))
print("Lift: " + str(item[2][0][3]))
```
