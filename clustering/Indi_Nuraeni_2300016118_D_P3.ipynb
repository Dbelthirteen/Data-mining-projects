**PERCOBAAN**

Import library dari Scikit-learn library

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
```

Menggunakan data artificial

```python
n_samples = 50
n_bins = 3
centers = [(-5, -5), (0, 0), (5, 5)]

X, _ = make_blobs(
    n_samples=n_samples,
    n_features=2,
    cluster_std=1.0,
    centers=centers,
    shuffle=False,
    random_state=42)

X
```

menampilkan sebaran data dengan memvisualkan dalam bentuk grafik

```python
plt.scatter(X[:, 0], X[:, 1])
plt.show()
```

Menjalankan metode k-mean untuk pengelompokan data

```python
KMeans_model = KMeans(n_clusters=3, random_state=1)
KMeans_model.fit(X)
```

Hasil output program



```python
kmeans = KMeans(
    algorithm='lloyd',
    copy_x=True,
    init='k-means++',
    max_iter=300,
    n_clusters=3,
    n_init=10,
    random_state=None,
    tol=0.0001,
    verbose=0)
```

Hasil pengclusteran

```python
label=KMeans_model.labels_
label
```

Percobaan dengan menggunakan data DATA_klas.xlxs

```python
data = pd.read_excel('DATA_klas.xlsx')
```

Menampilkan data

```python
data
```

Ukuran data

```python
data.shape
```

Mengambil nilai dari data

```python
records = []
for i in range(0, 11):
  records.append([int(data.values[i,j]) for j in range(2, 10)])
```

Transpose untuk melihat sebaran data

```python
tr=np.transpose(records)
tr
```

```python
plt.scatter(tr[0], tr[7])
plt.show()
```

Melihat tipe data

```python
type(records)
```

Analisis data dengan metode k-mean clustering

```python
KMeans_model =KMeans(n_clusters=2, random_state=1)
KMeans_model.fit(records)
```

Melihat label hasil pengelompokan

```python
labels = KMeans_model.labels_
labels
```

```python
KMeans_model = KMeans(n_clusters=2, random_state=1).fit(records)
labels = KMeans_model.labels_
labels
```
